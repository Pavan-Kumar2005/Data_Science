30. Build a menu-driven calculator that uses arithmetic, comparison, logical, assignment, membership, identity, and bitwise operators wherever applicable.

```python
def calculator():
    history = []  # used to demonstrate membership operator later

    while True:
        print("\n===== ALL-IN-ONE OPERATOR CALCULATOR =====")
        print("1. Arithmetic Operators (+, -, *, /, //, %, **)")
        print("2. Comparison Operators (==, !=, >, <, >=, <=)")
        print("3. Logical Operators (and, or, not)")
        print("4. Assignment Operators (+=, -=, *=, /=, etc.)")
        print("5. Membership Operators (in, not in)")
        print("6. Identity Operators (is, is not)")
        print("7. Bitwise Operators (&, |, ^, ~, <<, >>)")
        print("8. View Calculation History")
        print("9. Exit")

        choice = input("Enter your choice (1-9): ")

        # ---------------- 1. ARITHMETIC ----------------
        if choice == "1":
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            print(f"{a} + {b}  = {a + b}")
            print(f"{a} - {b}  = {a - b}")
            print(f"{a} * {b}  = {a * b}")
            if b != 0:
                print(f"{a} / {b}  = {a / b}")
                print(f"{a} // {b} = {a // b}")
                print(f"{a} % {b}  = {a % b}")
            else:
                print("Division/floor-division/modulus skipped (division by zero)")
            print(f"{a} ** {b} = {a ** b}")
            history.append(f"Arithmetic: {a}, {b}")

        # ---------------- 2. COMPARISON ----------------
        elif choice == "2":
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            print(f"{a} == {b} : {a == b}")
            print(f"{a} != {b} : {a != b}")
            print(f"{a} >  {b} : {a > b}")
            print(f"{a} <  {b} : {a < b}")
            print(f"{a} >= {b} : {a >= b}")
            print(f"{a} <= {b} : {a <= b}")
            history.append(f"Comparison: {a}, {b}")

        # ---------------- 3. LOGICAL ----------------
        elif choice == "3":
            p = input("Enter first condition (True/False): ").strip().lower() == "true"
            q = input("Enter second condition (True/False): ").strip().lower() == "true"
            print(f"{p} and {q} = {p and q}")
            print(f"{p} or {q}  = {p or q}")
            print(f"not {p}     = {not p}")
            history.append(f"Logical: {p}, {q}")

        # ---------------- 4. ASSIGNMENT ----------------
        elif choice == "4":
            x = float(input("Enter starting value: "))
            y = float(input("Enter value to apply in each operator: "))
            print(f"Starting value: {x}")
            x += y; print(f"x += {y}  -> {x}")
            x -= y; print(f"x -= {y}  -> {x}")
            x *= y; print(f"x *= {y}  -> {x}")
            if y != 0:
                x /= y; print(f"x /= {y}  -> {x}")
                x //= y; print(f"x //= {y} -> {x}")
                x %= y; print(f"x %= {y}  -> {x}")
            x **= 2; print(f"x **= 2   -> {x}")
            history.append(f"Assignment chain starting at {x}")

        # ---------------- 5. MEMBERSHIP ----------------
        elif choice == "5":
            value = input("Enter a value to search for: ")
            container = input("Enter comma-separated list to search in (e.g. a,b,c): ").split(",")
            container = [item.strip() for item in container]
            print(f"'{value}' in {container}     : {value in container}")
            print(f"'{value}' not in {container} : {value not in container}")
            history.append(f"Membership check: {value}")

        # ---------------- 6. IDENTITY ----------------
        elif choice == "6":
            list_a = [1, 2, 3]
            list_b = list_a          # same object
            list_c = [1, 2, 3]       # different object, same content
            print("list_a =", list_a, " list_b = list_a  list_c = [1,2,3] (new)")
            print(f"list_a is list_b     : {list_a is list_b}")
            print(f"list_a is list_c     : {list_a is list_c}")
            print(f"list_a is not list_c : {list_a is not list_c}")
            print(f"list_a == list_c     : {list_a == list_c}  (values match even though objects differ)")
            history.append("Identity demo run")

        # ---------------- 7. BITWISE ----------------
        elif choice == "7":
            a = int(input("Enter first integer: "))
            b = int(input("Enter second integer: "))
            print(f"{a} & {b}  (AND)         = {a & b}")
            print(f"{a} | {b}  (OR)          = {a | b}")
            print(f"{a} ^ {b}  (XOR)         = {a ^ b}")
            print(f"~{a}      (NOT)         = {~a}")
            print(f"{a} << 1  (Left Shift)  = {a << 1}")
            print(f"{a} >> 1  (Right Shift) = {a >> 1}")
            history.append(f"Bitwise: {a}, {b}")

        # ---------------- 8. HISTORY ----------------
        elif choice == "8":
            if history:
                print("\n--- Calculation History ---")
                for i, entry in enumerate(history, start=1):
                    print(f"{i}. {entry}")
            else:
                print("No calculations yet.")

        # ---------------- 9. EXIT ----------------
        elif choice == "9":
            print("Exiting calculator. Goodbye!")
            break

        else:
            print("Invalid choice. Please select a number between 1 and 9.")


calculator()
```
